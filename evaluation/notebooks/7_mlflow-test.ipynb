# MLFlow connection 

Connects to s3 minio storage and mlflow and saves a simple prompts from the console here.

```python
import mlflow
import subprocess
```

```python
mlflow.set_experiment("my-classification")
commit = subprocess.check_output(["git", "rev-parse", "HEAD"]).decode().strip()

with mlflow.start_run():
    mlflow.log_param("git_commit", commit)   # reproducibility
    mlflow.log_param("n_estimators", 100)
    mlflow.log_param("max_depth", 5)
    # model = train_model(X_train, y_train)
    mlflow.log_metric("accuracy", 0.94)
    mlflow.log_metric("f1_score", 0.91)
    # Critical: save the model, not just metrics
    # info = mlflow.sklearn.log_model(model, name="model")
```
