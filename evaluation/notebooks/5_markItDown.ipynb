```python
from src.vector_db.load_documents import load_file, load_documents
from langchain_core.documents import Document
```

```python
path = "../../data/styles/fiction/Book-Styles"

documents = load_documents(path)
documents
```

```python
from pathlib import Path


def save_documents(documents: list[Document], output_dir: str) -> None:
    """Write each converted Document to a .md file in output_dir."""
    out = Path(output_dir)
    out.mkdir(parents=True, exist_ok=True)

    for i, doc in enumerate(documents):
        # Reuse the original filename if load_file stored it in metadata
        source = doc.metadata.get("source")
        stem = Path(source).stem if source else f"document_{i}"
        (out / f"{stem}.md").write_text(doc.page_content, encoding="utf-8")
```

```python
save_documents(documents, "../../data/styles/fiction/Book-Styles/md")
```
